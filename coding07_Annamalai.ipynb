```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, LabelEncoder
```

```python
df=pd.read_csv(r"C:\Users\Anna1\OneDrive\UNCC Classes\DTSC 1301\CSV Datasets\student_survey.csv")
df.head()
```

```python
#One-hot encode transportation
transport_encoder= OneHotEncoder(sparse_output=False)

transport_encoded= transport_encoder.fit_transform(df[["transportation"]])
```

```python
transport_df= pd.DataFrame(transport_encoded,
columns = transport_encoder.get_feature_names_out(["transportation"])
)
```

```python
transport_df
```

```python
#Encode on campus

campus_encoder = OrdinalEncoder(categories=[["No","Yes"]])
```

```python
df["on_campus"]=campus_encoder.fit_transform(df[["on_campus"]])
```

```python
#Encode satisfaction

satisfaction_encoder= OrdinalEncoder(categories=[["Low","Medium","High"]])
df["satisfaction"]=satisfaction_encoder.fit_transform(df[["satisfaction"]])
```

```python
#Encode passed

passed_encoder = LabelEncoder()
df["passed"]=passed_encoder.fit_transform(df[["passed"]])
```

```python
#Remove original transportation column

df=df.drop(columns=["transportation"])
```

```python
#Add the encoded 
df= pd.concat([df.reset_index(drop=True), transport_df.reset_index(drop=True)],axis=0
)
```
